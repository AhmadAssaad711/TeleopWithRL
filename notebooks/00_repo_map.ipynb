# Repo Map

This notebook is the top-level entry point for the repo. Each code cell below renders the live repo structure and result roots inline.

```python
import sys
from pathlib import Path

for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / '.git').exists() and (candidate / 'matlab_env_python_replica').exists():
        if str(candidate) not in sys.path:
            sys.path.insert(0, str(candidate))
        break

from notebooks._teleop_nb import repo_paths, show_rows

P = repo_paths()
show_rows([
    {'name': name, 'path': str(path)}
    for name, path in P.items()
], title='Repo paths', max_rows=20)
```
