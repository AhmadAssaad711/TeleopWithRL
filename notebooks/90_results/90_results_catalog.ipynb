```python
import sys
from pathlib import Path

for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / '.git').exists() and (candidate / 'matlab_env_python_replica').exists():
        if str(candidate) not in sys.path:
            sys.path.insert(0, str(candidate))
        break

from notebooks._teleop_nb import repo_paths, count_tree, show_rows, show_csv, show_image

P = repo_paths()
raw_policy_results = P['repo'] / 'matlab_env_python_replica' / 'policy_gradient_experiments' / 'results'
show_rows([
    count_tree(raw_policy_results) | {'name': 'local policy-gradient results'},
    {'name': 'tracked result figures', 'root': str(P['results_index'] / 'figures'), 'dirs': 0, 'files': len(list((P['results_index'] / 'figures').glob('*.png')))},
], title='Current result inventory', max_rows=10)
```
