# Results Catalog

This notebook shows the three raw result roots and the generated flat `results_index/runs.csv` catalog inline.

```python
import sys
from pathlib import Path

for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / '.git').exists() and (candidate / 'matlab_env_python_replica').exists():
        if str(candidate) not in sys.path:
            sys.path.insert(0, str(candidate))
        break

from notebooks._teleop_nb import repo_paths, count_tree, show_rows

P = repo_paths()
roots = {
    'matlab_literal_results': P['matlab_results'],
    'dqn_results': P['dqn_results'],
    'ql_results': P['ql_results'],
}
show_rows([
    count_tree(root) | {'name': name}
    for name, root in roots.items()
], title='Result tree inventory', max_rows=10)
```

```python
from notebooks._teleop_nb import show_csv

show_csv(P['results_index'] / 'runs.csv', title='Indexed runs', max_rows=30)
```
