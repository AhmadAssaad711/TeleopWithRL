## Protocol and selection

The protocol is 500,000 PPO training steps, 32 test episodes, one train/evaluation signal pair, and 25 focused evaluation scenarios. The selected models are F4_accel_state, R5_second_order, and T3_posvel_current. F4 is the primary tracking candidate; R5 is the reward-design alternative; T3 is the temporal-observation alternative.

The current checkout has no TensorBoard event files. The training figure uses the saved `train.npz` export of the callback scalars (`teleop_train` and `teleop_eval`).

```python
import csv
import sys
from pathlib import Path

for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / '.git').exists() and (candidate / 'matlab_env_python_replica').exists():
        if str(candidate) not in sys.path:
            sys.path.insert(0, str(candidate))
        break

from notebooks._teleop_nb import repo_paths, show_rows, show_image

P = repo_paths()
selected_ids = {
    'physics_informed_formulations_03_fair_bias15_500k/F4_accel_state',
    'physics_reward_ablation_basic_obs_04_fair_bias15_500k/R5_second_order',
    'temporal_observation_stack_03_fair_bias15_500k/T3_posvel_current',
}
with (P['results_index'] / 'runs.csv').open(newline='', encoding='utf-8') as handle:
    rows = list(csv.DictReader(handle))
selected_rows = [row for row in rows if f"{row['study_id']}/{row['variant_id']}" in selected_ids]
show_rows(selected_rows, title='Selected model evaluation results', max_rows=10)
```
